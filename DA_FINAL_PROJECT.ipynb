##   Data Cleaning and Pre-processing



```python
# check duplicates
df.duplicated().sum()
```

```python
df.dtypes
```

```python
df["deposit_amount"]=pd.to_numeric(df["deposit_amount"])
```

```python
#column name cleaning
df.columns=df.columns.str.lower().str.strip()
```

```python
#categorical column
cat_cols = ["state_name","district_name","region","population_group"]

for col in cat_cols:
    df[col] = df[col].astype("category")
```

```python
# convert numpy array
data_array = df.to_numpy()
data_array[:5]
```

```python
#mean deposit using numpy
np.mean(df["deposit_amount"])
```

```python
# standard deviation
np.std(df["deposit_amount"])
```

```python
#find maximum deposit
np.max(df["deposit_amount"])
```

```python
#kurtosis checking
num_cols = ['no_of_offices', 'no_of_accounts', 'deposit_amount']

for col in num_cols:
    print(col, ":", df[col].kurt())
```

```python
#select numerical columns
num_cols=df.select_dtypes(include=np.number).columns
#IQR outliers detection
for col in num_cols:
  Q1=df[col].quantile(0.25)
  Q3=df[col].quantile(0.75)
  IQR=Q3-Q1
  lower=Q1-1.5*IQR
  upper=Q3+1.5*IQR
  outliers=df[(df[col]<lower)|(df[col]>upper)]
  print(col,":",len(outliers))
```

```python
# skewness checking
num_cols = ['no_of_offices', 'no_of_accounts', 'deposit_amount']

for col in num_cols:
    print(col, ":", df[col].skew())
```

* Skewness and kurtosis were calculated to understand the distribution pattern of the financial variables. The results indicate that the number of offices, number of accounts, and deposit amount are highly positively skewed. This means that while most districts exhibit moderate banking activity, a small number of districts have significantly higher values.

* Kurtosis values are also high and positive, indicating heavy-tailed distributions and the presence of extreme values. This confirms that the dataset contains strong outliers, particularly in financial variables.

* The IQR method was applied to identify outliers. Although a considerable number of extreme values were detected, they were retained in the dataset as they represent genuine economic and banking variations across districts.

##  EDA and Visualizations



## Univariate analysis:

```python
#univariate analysis
#count plot for categorical coulumn

plt.figure(figsize=(10,5))
sns.countplot(x='state_code', data=df)
plt.title("Count of Records by State Code")
plt.xticks(rotation=90)
plt.show()
```

**Insights :**

The count plot shows the state_code 9 has highest number of district records.This suggests that Utter pradesh has wider district coverage or more recorded banking data compared to other states.

**Interpretation :**

The count plot of state_code shows the distribution of district-level records across different states. Certain states have a higher number of districts recorded, while others have fewer entries. This indicates variation in administrative or geographical coverage across states.

```python
#Univariate analysis
#Histogram for Numerical columns

plt.figure()
df["deposit_amount"].plot(kind="hist", bins=20)
plt.title("Distribution of Deposit Amount")
plt.xlabel("Deposit Amount")
plt.ylabel("Frequency")
plt.show()
```

### **Insights**

* The deposit amount is Positively skewed.
* Most districts have low to medium deposit amount.
* Few districts have very high deposits.


## Interpretation of your above chart
The histogram of deposit amount reveals a highly right-skewed distribution. Most districts record moderate deposit levels, whereas a small number of economically developed districts contribute disproportionately higher deposits.


```python
#Univariate analysis
#Boxplot for outlier visualization

plt.figure()
sns.boxplot(x=df['no_of_accounts'])
plt.title("Number of Accounts")
plt.xlabel("No of accounts")
plt.show()

```

**Insights:**

* Most districts have moderate number of accounts
* It reflects unequal banking distribution.


## Interpretation of your above chart
* The box plot of number of accounts shows that the median lies closer to the lower quartile, indicating a positively skewed distribution.
* The interquartile range (IQR) captures the middle 50% of districts with moderate account values.
* Several data points appear beyond the upper whisker, indicating the presence of extreme values (outliers).

## Bivariate analysis

```python
#Bivariate analysis
#Scatter plot
plt.figure(figsize=(8,5))
sns.scatterplot(x='no_of_accounts', y='deposit_amount', data=df)

plt.title('Scatterplot: No of Accounts vs Deposit Amount')
plt.xlabel('Number of Accounts')
plt.ylabel('Deposit Amount')
plt.show()
```

**Insights**
* Districts with more customers generate higher deposit volumes, proving customer base directly impacts revenue.
* Expanding account openings in low-performing districts could significantly increase total deposits.
*  High-performing districts can be studied as benchmark regions to replicate growth strategies in other areas.

## Interpretation of your above chart
* The scatterplot shows an upward trend, indicating a positive relationship between number of accounts and deposit amount.
*  Most data points follow a linear pattern, suggesting consistent growth in deposits as accounts increase.
* A few districts appear far from the cluster, indicating possible high-value outliers.


```python
# Bivariate analysis
# Bar plot
plt.figure(figsize=(10,5))
sns.barplot(x='state_code', y='deposit_amount', data=df)

plt.title('Barplot: State Code vs Deposit Amount')
plt.xlabel('State Code')
plt.ylabel('Average Deposit Amount')
plt.show()
```

**Insights:**
* Certain states are major contributors to total deposits, indicating strong regional economic activity.
* Low-performing states may require targeted marketing or financial inclusion programs.
* The bank can allocate resources strategically based on high and low deposit-performing states.

**Interpretation:**
* The bar plot shows variation in average deposit amount across different state codes.
* Some states have significantly higher deposit values compared to others.
* There is noticeable imbalance, indicating uneven financial performance across regions

```python
#Bivariate analysis
#correlation heatmap
plt.figure(figsize=(8,5))
correlation = df[['no_of_accounts', 'deposit_amount']].corr()

sns.heatmap(correlation, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

**Insights:**
* Increasing customer acquisition will directly increase deposit growth.
* The bank should focus on expanding its customer base to improve financial performance.
* Customer engagement strategies can significantly impact total deposit

## Interpretation of your above chart
* The correlation value between number of accounts and deposit amount is strongly positive.
* The closer the value is to +1, the stronger the linear relationship.
* This indicates that both variables move in the same direction.
